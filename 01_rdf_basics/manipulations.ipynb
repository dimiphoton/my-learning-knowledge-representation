# Étape 1 — Manipulations : construire un graphe RDF triplet par triplet

Ce notebook accompagne le `README.md`. On va **construire le graphe à la main, un triplet à la fois**, avec `rdflib`, pour bien sentir la mécanique Sujet → Prédicat → Objet.

Exécute les cellules **dans l'ordre**, de haut en bas (Maj+Entrée).

## 1. Un graphe vide

Un graphe RDF est juste un **sac de triplets**. On commence par un sac vide. `len(graphe)` compte le nombre de triplets : il vaut 0 pour l'instant.

````python
from rdflib import Graph

graphe = Graph()
print(f"Nombre de triplets dans le graphe : {len(graphe)}")
````

## 2. Nommer les choses avec des IRIs

Avant d'écrire un triplet, il faut des **noms uniques** pour nos sujets, prédicats et objets.

- `URIRef("...")` = une IRI (un identifiant global), l'équivalent de `<...>` en Turtle.
- `Literal("...")` = une valeur brute (texte, nombre), l'équivalent de `"..."` en Turtle.

Pour éviter de réécrire `http://exemple.org/` à chaque fois, on crée un raccourci avec `Namespace` (c'est l'ancêtre Python du `@prefix`).

````python
from rdflib import Namespace, Literal
from rdflib.namespace import RDF  # RDF.type = le predicat "est de type"

# EX est notre espace de noms : EX.alice == URIRef("http://exemple.org/alice")
EX = Namespace("http://exemple.org/")

# Les sujets/objets (des IRIs)
alice = EX.alice
bob = EX.bob
personne = EX.Personne

# Le predicat "connait" (une IRI aussi)
connait = EX.connait

print("alice     ->", alice)
print("RDF.type  ->", RDF.type)
print("personne  ->", personne)
````

## 3. Ajouter notre tout premier triplet

On ajoute un triplet avec `graphe.add((sujet, predicat, objet))`. Attention aux **doubles parenthèses** : `add()` prend **un seul argument** qui est un tuple `(s, p, o)`.

Premier triplet : « **Alice est une Personne** » → `(alice, RDF.type, personne)`.

````python
# "Alice est une Personne"
graphe.add((alice, RDF.type, personne))

print(f"Nombre de triplets : {len(graphe)}")
````

## 4. Ajouter les autres triplets

On enrichit le graphe : Bob est une Personne, Alice connaît Bob, et chacun a un `nom` (un **littéral** cette fois, donc `Literal(...)`).

````python
nom = EX.nom

graphe.add((bob, RDF.type, personne))          # Bob est une Personne
graphe.add((alice, connait, bob))               # Alice connait Bob
graphe.add((alice, nom, Literal("Alice")))      # nom d'Alice (litteral)
graphe.add((bob, nom, Literal("Bob")))          # nom de Bob (litteral)

print(f"Nombre de triplets : {len(graphe)}")
````

## 5. Voir le graphe au format Turtle

`graphe.serialize(format="turtle")` transforme notre sac de triplets en texte Turtle lisible. Tu devrais reconnaître le contenu de `personnes.ttl` (en version raccourcie par les préfixes que rdflib génère automatiquement).

````python
print(graphe.serialize(format="turtle"))
````

## 6. Parcourir le graphe en Python

On peut boucler sur les triplets. Chaque triplet est un tuple `(sujet, predicat, objet)`. Ici on n'affiche que les relations « connaît ».

````python
# Le 2e argument None signifie "n'importe quel predicat" ; ici on fixe le predicat 'connait'.
for sujet, predicat, objet in graphe.triples((None, connait, None)):
    print(f"{sujet} connait {objet}")
````
